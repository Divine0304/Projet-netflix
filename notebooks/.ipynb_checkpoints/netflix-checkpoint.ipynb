```python
# Fusion et calcul des points
resultat_final = pd.concat([df_final, df_manuels], ignore_index=True)

resultat_final['points'] = resultat_final['difficulty'].map({
    'Évident/culte': 5,
    'Facile/accessible': 10,
    'Moyen/technique': 20,
    'Difficile/expert': 30
})

print(f"Total de la liste : {len(resultat_final)} films.")
```

```python
import json

#  : Génération du fichier data/films.js
json_data = resultat_final.to_dict(orient='records')
with open('../data/films.js', 'w', encoding='utf-8') as f:
    f.write(f"const filmsData = {json.dumps(json_data, ensure_ascii=False, indent=4)};")

print("Fichier data/films.js généré ! ")
```
